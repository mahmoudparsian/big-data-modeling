# 📓 PySpark + Jupyter: A Simple Example

This notebook creates a small PySpark DataFrame and filters it —
the same example used in [`../marimo/pyspark_marimo_example.py`](../marimo/pyspark_marimo_example.py),
so you can compare the two notebook styles side by side.

**Setup:**
```bash
pip install jupyter pyspark
jupyter notebook pyspark_jupyter_example.ipynb
```

Unlike the marimo version, this notebook is **not reactive**: if you
change `MIN_SALARY` below, you have to manually re-run that cell (and
any cell after it) yourself — Jupyter won't do it for you.

## Step 1 — Start a SparkSession

````python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = (
    SparkSession.builder
    .appName("pyspark-jupyter-example")
    .master("local[*]")
    .config("spark.ui.showConsoleProgress", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")
````

## Step 2 — Create a Sample DataFrame

````python
some_data = [
    ("alex", 20, 12000),
    ("jane", 30, 45000),
    ("rafa", 40, 56000),
    ("ted",  30, 145000),
    ("xo2",  10, 1332000),
    ("mary", 44, 555000),
]
column_names = ["name", "age", "salary"]
df = spark.createDataFrame(some_data, column_names)
df.show()
````

## Step 3 — Filter

Change `MIN_SALARY` and re-run this cell to see a different result
(there's no slider here — that's the trade-off for not needing a
running reactive kernel).

````python
MIN_SALARY = 50_000

filtered = df.filter(col("salary") >= MIN_SALARY)
filtered.show()
````

## Step 4 — Stop Spark

````python
spark.stop()
````
